FEATURE SCALING

```python
df = pd.read_csv('/content/Social_Network_Ads.csv', usecols=[2,3,4])
df.head()
```

```python
x = df.drop(['Purchased'], axis = 1)
y = df['Purchased']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.3, random_state = 0)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
x_train.head(5)
```

```python
# STANDARD SCALER
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

# Fit the Train set, it will learn the parameters
scaler.fit(x_train)

# Transform train and test sets
x_train_scaled = scaler.transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
# x_train = Dataframe
# After scaling x_train_scaled becomes Numpy array
# Convert to dataframe first

x_train_scaled = pd.DataFrame(x_train_scaled, columns = x_train.columns)
x_test_scaled = pd.DataFrame(x_test_scaled, columns = x_test.columns)
```

```python
# Checking -> Before scaling
np.round(x_train.describe(),1)
```

```python
# Checking -> After Scaling
np.round(x_train_scaled.describe(),1)
```

```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("/content/customer.csv", usecols = [2,3,4])
df.head()
```

```python
x = df.drop('purchased', axis =1)
y = df['purchased']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.2)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
# Ordinal Encoding
from sklearn.preprocessing import OrdinalEncoder
ode = OrdinalEncoder(categories = [['Poor','Average','Good'],['School','UG','PG']])

ode.fit(x_train)

x_train_scaled = ode.transform(x_train)
x_test_scaled = ode.transform(x_test)
```

```python
x_train_scaled = pd.DataFrame(x_train_scaled, columns = x_train.columns)
x_test_scaled = pd.DataFrame(x_test_scaled, columns = x_test.columns)
```

```python
x_train.head(4)
```

```python
x_train_scaled.head(4)
```

```python
# Label Encoding
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()

le.fit(y_train)

y_train_scaled = le.transform(y_train)
y_test_scaled = le.transform(y_test)
```
