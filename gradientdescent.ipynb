```python
from sklearn.datasets import make_regression
import numpy as np
```

```python
x,y = make_regression(n_samples = 4, n_features=1, n_informative=1, n_targets = 1, noise=80, random_state = 13)
print(x)
print(y)
```

```python
import matplotlib.pyplot as plt
plt.scatter(x,y)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
regg = LinearRegression()
regg.fit(x,y)
```

```python
regg.coef_
```

```python
regg.intercept_
```

```python
plt.scatter(x,y)
plt.plot(x,regg.predict(x), color = "red")
```

```python
# Apply GD assuming slope is constant and let intercept is zero
y_pred = ((78.35 * x)+0).reshape(4)
```

```python
plt.scatter(x,y)
plt.plot(x,regg.predict(x), color = "red")
plt.plot(x,y_pred, color = "green")
```

```python
m = 78.35
b = 0
loss_slope = -2 * np.sum(y - m*x.ravel() - b)
loss_slope
```

```python
# Let Learning rate = 0.1
lr = 0.1
step_size = loss_slope * lr
step_size
```

```python
# Calculating the new intercept
b = b - step_size
b
```

```python
y_pred1 = ((78.35 * x)+b).reshape(4)
plt.scatter(x,y)
plt.plot(x,regg.predict(x), color = "red")
plt.plot(x,y_pred, color = "green")
plt.plot(x,y_pred1, color = "black")
```

```python
# Interation 2
loss_slope = -2 * np.sum(y - m*x.ravel() - b)
loss_slope
```

```python
lr = 0.1
step_size = loss_slope * lr
step_size
```

```python
b = b - step_size
b
```

```python
y_pred2 = ((78.35 * x)+b).reshape(4)
plt.scatter(x,y)
plt.plot(x,regg.predict(x), color = "red")
plt.plot(x,y_pred, color = "green")
plt.plot(x,y_pred1, color = "black")
plt.plot(x,y_pred2, color = "yellow")
```

```python
# Iteration 3
loss_slope = -2 * np.sum(y - m*x.ravel() - b)
loss_slope
```

```python
lr = 0.1
step_size = loss_slope * lr
step_size
```

```python
b = b - step_size
b
```

```python
y_pred3 = ((78.35 * x)+b).reshape(4)
plt.scatter(x,y)
plt.plot(x,regg.predict(x), color = "red")
plt.plot(x,y_pred, color = "green")
plt.plot(x,y_pred1, color = "black")
plt.plot(x,y_pred2, color = "yellow")
plt.plot(x,y_pred3, color = "grey")
```

```python
# In Loop
m = 78.35
b = 0
lr = 0.1
epochs = 100
for i in range(epochs):
  loss_slope = -2 * np.sum(y - m*x.ravel() - b)
  b = b - (loss_slope * lr)
  y_pred = m*x + b

  plt.plot(x,y_pred)
plt.scatter(x,y, color = "green")
```
