```python
import numpy as np
import pandas as pd
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
```

```python
df = pd.read_csv('/content/enhanced_fever_medicine_recommendation.csv', usecols = [2,3,1,18,17])
df.columns = ['Fever_Severity','Age','Gender','Blood_Pressure','Previous_Medication']
df= df[['Age','Gender','Fever_Severity','Blood_Pressure','Previous_Medication']]
df.head()
```

```python
df['Blood_Pressure'].unique()
```

```python
df.isnull().sum()
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(df.drop('Previous_Medication',axis = 1), df['Previous_Medication'], test_size = 0.2)
```

Without columns transform (long process)

```python
# for treating missing values of previous medication
si = SimpleImputer()
y_train_medication = si.fit_transform(y_train[['Previous_Medication']])
y_test_medication = si.fit_transform(y_test[['Previous_Medication']])

y_train_medication
```

```python
# for oe for gender
oe = OrdinalEncoder(categories =[['Male','Female']])
x_train_gender = oe.fit_transform(x_train[['Gender']])
x_test_gender = oe.fit_transform(x_test[['Gender']])
x_train_gender
```

```python
# ohe for fever and blood pressure
ohe = OneHotEncoder(drop = 'first',sparse_output = False)
x_train_fever_bp = ohe.fit_transform(x_train[['Fever_Severity','Blood_Pressure']])
x_test_fever_bp = ohe.fit_transform(x_test[['Fever_Severity','Blood_Pressure']])
x_train_fever_bp
```

```python
#extracting of age
x_train_age = x_train.drop(columns = ['Fever_Severity','Blood_Pressure','Gender']).values
x_test_age = x_test.drop(columns = ['Fever_Severity','Blood_Pressure','Gender']).values
```

```python
#joining all
x_train_transform = np.concatenate((x_train_age,x_train_fever_bp,x_train_gender), axis =1)
x_test_transform = np.concatenate((x_test_age,x_test_fever_bp,x_test_gender), axis =1)
x_train_transform
```

With column transformer

```python
from sklearn.compose import ColumnTransformer
```

```python
transformer = ColumnTransformer(transformers=[
    ('tnf1', OneHotEncoder(sparse_output = False, drop = 'first'), ['Fever_Severity','Blood_Pressure']),
    ('tnf2',OrdinalEncoder(categories=[['Male','Female']]),['Gender'])
],remainder='passthrough')
```

```python
transformer.fit_transform(x_train)
```

```python
transformer.transform(x_test)
```
