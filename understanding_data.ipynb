```python
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv("/content/Titanic-Dataset.csv")
```

```python
# How big is the data?
df.shape
```

```python
# How does data look like?
df.head()
df.sample(5)
```

```python
# What is datatype of columns?
df.info()
```

```python
# How many missing values are there?
df.isnull().sum()
```

```python
# How does the data look mathematically?
df.describe()
```

```python
# Are there any duplicate values?
df.duplicated('PassengerId').sum()
```

```python
# How is correlation between columns?
df.corr(numeric_only = True)['Survived']
```

```python
# Univariate Analysis
# Categorical data
# 1. Count plot
sns.countplot(x = df['Survived'], color = 'red') #/ #sns.countplot(x = 'Survived', data = df)
df['Survived'].value_counts().plot(kind = 'bar')
# 2. Pie chart
df['Survived'].value_counts().plot(kind = 'pie', autopct = '%0.2f')
```
