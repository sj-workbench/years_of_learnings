```python
import pandas as pd
```

```python
# Series
```

```python
# String
countries = ['india', 'usa', ' japan','nepal','south korea']
pd.Series(countries[2])
```

```python
# Integer
runs = [32,100,22,1000,567]
pd.Series(runs[3])
```

```python
# Custom index
gender = ["male","female","male","female","male"]
genders = pd.Series(gender,index=["a","b","c","d","e"])
genders
```

```python
# setting a name
gender_name = pd.Series(gender,index=["a","b","c","d","e"], name = 'GENDER')
```

```python
# series from dict
marks = {
    'math':67,
    'english':57,
    'odia': 99,
    "japanese": 37
}
mark_series = pd.Series(marks, name ="MARKS")
mark_series
```

```python
# series attributes
# size
mark_series.size  # counts missing values
# dtype
genders.dtype
# name
gender_name.name
# is unique
gender_name.is_unique  # gives repetation of items as true or false
# index
mark_series.index
# values
mark_series.values
gender_name.values[0]
```

```python
# series using read_csv
# with one column
data = pd.read_csv("/content/car data.csv").squeeze()
car_name = data['Car_Name']
type(car_name)
car_name.head()
# two columns
placement = pd.read_csv("/content/placement.csv", index_col = 'cgpa').squeeze()
placement.head()
```

```python
# Series methods
# head and tail
placement.head()
car_name.tail()
# sample
placement.sample(5)
# value counts
placement.value_counts().sample()
# sort_values
placement.sort_values(ascending=False).head()   # ascending and descending available, default = True
# sort_index
car_name.sort_index()
```

```python
# Series maths method
# count
placement.count()  # doesnt count missing values
# sum
placement.sum()
# product
placement.product()
# mean
placement.mean()
# median
placement.median()
# std
placement.std()
# min
placement.min()
# max
placement.max()
# describe
placement.describe()     # gives a overview data of data
```
