```python
class GDregressor:
  def __init__(self,learning_rate=0.0001,epochs=10):
    self.m = 26.08
    self.b = -120
    self.learning_rate = learning_rate
    self.epochs = epochs
  def fit(self,x,y):
    # Calculating GD
    n = len(x)
    for i in range(self.epochs):
      loss_slope = -(2/n)* np.sum(y-self.m*x.ravel() - self.b)
      self.b = self.b - (self.learning_rate*loss_slope)
      print(self.b)
```

```python
gd = GDregressor()
gd.fit(x,y)
```
