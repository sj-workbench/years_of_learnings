```python
import pandas as pd
```

```python
df = pd.read_csv('/content/archive (6).zip')
print(df.head())
```

```python
print(df.shape)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(df['mileage'], df['price'])
```

```python
X = df [['year', 'mileage']]
Y = df ['price']
```

```python
print(X.head())
```

```python
print(Y.head())
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size = 0.2, random_state = 42)
```

```python
print(len(X_train))
print(len(X_test))
```

```python
print(X_train)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
clsf = LinearRegression()
```

```python
clsf.fit(X_train, Y_train)
```

```python
clsf.predict(X_test)
```

```python
print(Y_test)
```

```python
clsf.score(X_test,Y_test)
```
