```python
# series using read_csv
# with one column
data = pd.read_csv("/content/car data.csv").squeeze()
car_name = data['Car_Name']
type(car_name)
car_name.head()
# two columns
placement = pd.read_csv("/content/placement.csv", index_col = 'cgpa').squeeze()
placement.head()
```

```python
# Series indexing
```

```python
numbers = [12,23,45,64,'SJ']
numbers = pd.Series(numbers, index = ['a','b','c','d','e'], name = "AGE")
numbers[-1]                        # -ve/+ve works
```

```python
number = pd.Series([12,'SJ',42,43,21], name = 'House')
number[1]                        # +ve only works
```

```python
# Series slicing
place = ['japan','south korea', 'LA','London', 'India','Spain', 'Europe','Thailand']
places = pd.Series(place, index = [1,2,3,4,5,6,7,8])
places
place[5::2]
place[::2]
```

```python
# fancy indexing
places[[1,4,5]]
```

```python
# Editng item inside series
# using indexing
marks = [23,45,65,78,99,76,54]
marks = pd.Series(marks, name = 'mark')
marks[1] = 100   # edited
marks
# what if an index doesnt not exist
marks[7] = 288
marks
# with slicing
marks[2:5:2] = [75,89]
marks
# with fancy indexing
marks[[0,1,2]] = [0,0,0]
marks
```

```python
# Series with python functionalities
# type/dir/len/sorted/max/min
print(len(marks))
print(type(marks))
print(dir(marks))
print(sorted(marks))
print(min(marks))
print(max(marks))

# type conversion
list(marks)
dict(marks)

# membership operator
9 in marks     # used index only
76 in marks.values   # used for values

# loop
for i in marks.index:
  print(i)        # runs on values only but for index use .index

# operators
# Arithmetic operatorss
100 - marks   # example of broadcasting
new_marks = marks + 2
new_marks
# relational operators
marks<10    # boolean series
marks>100
```

```python
# plotting on series
marks.plot()
marks.plot(kind = 'bar')
marks.plot(kind = 'pie')
```
