```python
import numpy as np
import pandas as pd
```

```python
covid = pd.read_csv("/content/covid_toy.csv")
covid.head()
```

```python
covid['cough'].value_counts()
```

```python
covid['city'].value_counts()
```

```python
covid.isnull().sum()
```

```python
x = covid.drop('has_covid', axis = 1)
y = covid['has_covid']
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size = 0.2, random_state=0)
```

```python
# Column transformer
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder
from sklearn.impute import SimpleImputer
```

```python
transformer = ColumnTransformer(transformers = [
    ('transformer1', SimpleImputer(), ['fever']),
    ('transformer2', OrdinalEncoder(categories = [['Mild','Strong']]),['cough']),
    ('transformer3', OneHotEncoder(drop = 'first', sparse_output = False), ['gender','city']),
],
                              remainder = 'passthrough')
```

```python
(transformer.fit_transform(x_train))
```

```python
transformer.fit_transform(x_train).shape
```

```python
transformer.transform(x_test).shape
```
