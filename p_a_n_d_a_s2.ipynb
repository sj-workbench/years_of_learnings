```python
import pandas as pd
```

```python
# Series
```

```python
# String
countries = ['india', 'usa', ' japan','nepal','south korea']
pd.Series(countries[2])
```

```python
# Integer
runs = [32,100,22,1000,567]
pd.Series(runs[3])
```

```python
# Custom index
gender = ["male","female","male","female","male"]
genders = pd.Series(gender,index=["a","b","c","d","e"])
genders
```

```python
# setting a name
gender_name = pd.Series(gender,index=["a","b","c","d","e"], name = 'GENDER')
```

```python
# series from dict
marks = {
    'math':67,
    'english':57,
    'odia': 99,
    "japanese": 37
}
mark_series = pd.Series(marks, name ="MARKS")
mark_series
```

```python
# series attributes
# size
mark_series.size  # counts missing values
# dtype
genders.dtype
# name
gender_name.name
# is unique
gender_name.is_unique  # gives repetation of items as true or false
# index
mark_series.index
# values
mark_series.values
gender_name.values[0]
```

```python
# series using read_csv
# with one column
data = pd.read_csv("/content/car data.csv").squeeze()
car_name = data['Car_Name']
type(car_name)
car_name.head()
# two columns
placement = pd.read_csv("/content/placement.csv", index_col = 'cgpa').squeeze()
placement.head()
```

```python
# Series methods
# head and tail
placement.head()
car_name.tail()
# sample
placement.sample(5)
# value counts
placement.value_counts().sample()
# sort_values
placement.sort_values(ascending=False).head()   # ascending and descending available, default = True
# sort_index
car_name.sort_index()
```

```python
# Series maths method
# count
placement.count()  # doesnt count missing values
# sum
placement.sum()
# product
placement.product()
# mean
placement.mean()
# median
placement.median()
# std
placement.std()
# min
placement.min()
# max
placement.max()
# describe
placement.describe()     # gives a overview data of data
```

```python
# Series indexing
```

```python
numbers = [12,23,45,64,'SJ']
numbers = pd.Series(numbers, index = ['a','b','c','d','e'], name = "AGE")
numbers[-1]                        # -ve/+ve works
```

```python
number = pd.Series([12,'SJ',42,43,21], name = 'House')
number[1]                        # +ve only works
```

```python
# Series slicing
place = ['japan','south korea', 'LA','London', 'India','Spain', 'Europe','Thailand']
places = pd.Series(place, index = [1,2,3,4,5,6,7,8])
places
place[5::2]
place[::2]
```

```python
# fancy indexing
places[[1,4,5]]
```

```python
# Editng item inside series
# using indexing
marks = [23,45,65,78,99,76,54]
marks = pd.Series(marks, name = 'mark')
marks[1] = 100   # edited
marks
# what if an index doesnt not exist
marks[7] = 288
marks
# with slicing
marks[2:5:2] = [75,89]
marks
# with fancy indexing
marks[[0,1,2]] = [0,0,0]
marks
```

```python
# Series with python functionalities
# type/dir/len/sorted/max/min
print(len(marks))
print(type(marks))
print(dir(marks))
print(sorted(marks))
print(min(marks))
print(max(marks))

# type conversion
list(marks)
dict(marks)

# membership operator
9 in marks     # used index only
76 in marks.values   # used for values

# loop
for i in marks.index:
  print(i)        # runs on values only but for index use .index

# operators
# Arithmetic operatorss
100 - marks   # example of broadcasting
new_marks = marks + 2
new_marks
# relational operators
marks<10    # boolean series
marks>100
```

```python
# plotting on series
marks.plot()
marks.plot(kind = 'bar')
marks.plot(kind = 'pie')
```
