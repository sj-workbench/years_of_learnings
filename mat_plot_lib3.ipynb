```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
android = [55000,58000,61000,63000,75000,95000]
years = [2012,2013,2014,2015,2016,2017]
yearss = [2012,2013,2014,2015,2016,2017]
iphone = [55000,100000,120000,125000,175000,200000]
plt.xlabel("Years")
plt.ylabel("prices")
plt.title("android vs iphone")
# label on lines
plt.plot(years,iphone, label = 'iphone')
plt.plot(yearss,android, label = 'android')
# loc = managing location of label
plt.legend(loc = 'upper center')
# limiting axes
plt.xlim(2012,2015)
plt.ylim(40000,100000)
# grid
plt.grid()
plt.show()
```

```python
batsman = pd.read_csv("/content/Virat Kohli DataSet.csv")
batsman
# color, linestyle, plotting, width, marker, marker size
plt.plot(batsman["Runs"], batsman["Opposition"], color = 'green', linestyle= 'dashed', linewidth = 2, marker = '+', markersize = 10)
# title
plt.title("`Runs VS Opposition`")
# labeling
plt.xlabel("Runs")
plt.ylabel("Opposition")
plt.show()
# multiple plot
#plt.plot(batsman["Runs"], batsman["BF"], color = "green")    #can plot multiple on a single graph
```

```python
# pie chart
data = [10,20,30,55,69,99]
subjects = ['eng','maths','sci','evs','phy','chem']
plt.pie(data, labels = subjects, autopct = "%0.1f%%", colors = ['blue','green','yellow','pink', 'cyan', 'brown'] , explode = [0,0,0,0,0,0.09], shadow = True)
plt.show()
```
