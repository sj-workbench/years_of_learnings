```python
from sklearn.datasets import make_regression
from sklearn.linear_model import LinearRegression
import pandas as pd
import numpy as np

import plotly.express as px
import plotly.graph_objects as go

from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
```

```python
x,y = make_regression(n_samples= 100, n_features=2, n_informative = 2, n_targets = 1, noise=20)
```

```python
df = pd.DataFrame({'f1':x[:,0],'f2':x[:,1],'target':y})
```

```python
df.head()
```

```python
fig = px.scatter_3d(df,x='f1',y='f2',z='target')
fig.update_traces(marker=dict(size = 5, color = 'red'))
fig.show()
```

```python
df.shape
```

```python
x = df.iloc[:,0:2]
y = df.iloc[:,-1]
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=3)
```

```python
x_train.head()
```

```python
lr = LinearRegression()
lr.fit(x_train,y_train)
```

```python
y_pred = lr.predict(x_test)
```

```python
acc = r2_score(y_test,y_pred)
acc1 = mean_absolute_error(y_test,y_pred)
acc2 = mean_squared_error(y_test,y_pred)
print(acc)
print(acc1)
print(acc2)
```
