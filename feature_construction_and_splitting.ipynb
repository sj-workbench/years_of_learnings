```python
import numpy as np
import pandas as pd
import seaborn as sns

from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LogisticRegression
```

```python
df = pd.read_csv('/content/Titanic-Dataset.csv', usecols = {1,2, 5,6,7})
df
```

```python
df.dropna(inplace=True)
```

```python
df
```

```python
x= df.iloc[:,1:5]
y= df.iloc[:,-5]
```

```python
x.head()
```

```python
y.head()
```

```python
np.mean(cross_val_score(LogisticRegression(), x, y, scoring='accuracy', cv=20))
```

```python
# Applying Feature Contruction
x['Family'] = x['SibSp']+x['Parch']+1
```

```python
x.head()
```

```python
def myfunc(num):
  if num==1:
    #alone
    return 0

  elif num >1 and num<=4:
    #small family
    return 1

  else:
    return 2
```

```python
x['Family_type']= x['Family'].apply(myfunc)
```

```python
x.head()
```

```python
x.drop(columns = ['SibSp','Parch','Family'],inplace = True)
```

```python
x.head()
```

```python
np.mean(cross_val_score(LogisticRegression(), x, y, scoring='accuracy', cv=20))
```
